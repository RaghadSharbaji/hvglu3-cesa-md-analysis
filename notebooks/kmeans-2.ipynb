```python
%matplotlib inline
import contact_map
import matplotlib.pyplot as plt
from matplotlib import colormaps
import mdtraj as md
import numpy as np
traj = md.load("S1_NpHLpH_all.xtc", top="backbone.pdb")
topology = traj.topology
from contact_map import ContactFrequency, ContactDifference, ContactTrajectory, RollingContactFrequency
print(traj)
```

```python
import scipy.stats as st
import pandas as pd
import pyemma
import seaborn as sns; sns.set()  # for plot styling
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.metrics import make_scorer
from sklearn.datasets import make_classification
my_scorer = make_scorer(mean_squared_error, greater_is_better=False)
```

```python
# Perform PCA
torsions_feat=pyemma.coordinates.featurizer(topology)
torsions_feat.add_distances_ca("all")
traj=("S1_NpHLpH_all.xtc")
torsions_data=pyemma.coordinates.load(traj, features= torsions_feat)
labels = ['backbone\torsions']
data1 = PCA(n_components=3).fit_transform(torsions_data)
data1 = pd.DataFrame(data1, columns=['f','g','i']) # f:PC1, g:PC2, i:PC3
import matplotlib.pyplot as plt

# Plot the first two principal components
plt.figure(figsize=(8,6))
plt.scatter(data1['f'], data1['g'], alpha=0.6)
plt.xlabel('PC1 (f)')
plt.ylabel('PC2 (g)')
plt.title('PCA of Torsion Features')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

from sklearn.cluster import KMeans

# Use PCA-transformed data for clustering
features = data1[['f', 'g', 'i']].values

# Optionally standardize (good practice)
from sklearn.preprocessing import StandardScaler
features_scaled = StandardScaler().fit_transform(features)

# Elbow method
inertia = []
k_range = range(1, 11)
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=0)
    kmeans.fit(features_scaled)
    inertia.append(kmeans.inertia_)

# Plot the elbow
plt.figure(figsize=(8,6))
plt.plot(k_range, inertia, 'bo-', linewidth=2)
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal K')
plt.xticks(k_range)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
kmeans = KMeans(n_clusters=5) # Specify number of clusters
kmeans.fit(data1)
y_kmeans = kmeans.predict(data1)

plt.scatter(data1["f"], data1["g"], c=y_kmeans, s=50, cmap='viridis') # PC1 vs. PC2

centers = kmeans.cluster_centers_ # Centroids

plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.5);
# plt.xlim(-5,8)
# plt.ylim(-6,7.5)
plt.title('K-means Clustering S1',weight='bold',fontsize=18)
plt.xlabel('PCA-1',fontsize=14)
plt.ylabel('PCA-2',fontsize=14)
plt.show()
```
